```python
import numpy, scipy, statistics, math
y1 = [64, 70, 72, 74, 81, 83, 87, 90, 94, 96, 104, 119]
y2 = [82, 91, 103, 118, 142, 175]
n1 = len(y1)
n2 = len(y2)
s2 = ( (n1-1)*statistics.variance(y2)+(n2-1)*statistics.variance(y1) ) / (n1+n2-2)
tequal = ( statistics.mean(y1) - statistics.mean(y2) ) / math.sqrt( s2 *(1/n1+1/n2) )
tcritequal = scipy.stats.t.ppf( 0.975, n1+n2-2)
pvalequal = 2*(1 - scipy.stats.t.cdf ( abs(tequal), n1+n2-2))
print(tequal)
print(tcritequal)
print(pvalequal)
se12 = statistics.variance(y1)/n1
se22 = statistics.variance(y2)/n2
df1 = n1-1
df2 = n2-1
df = (se12+se22)*(se12+se22) / (se12*se12/df1 + se22*se22/df2)
tunequal = ( statistics.mean(y1) - statistics.mean(y2) ) / math.sqrt( se12 + se22 )
tcritunequal = scipy.stats.t.ppf (.975, df)
pvalunequal = 2*(1 - scipy.stats.t.cdf ( abs(tunequal), df))
print(tunequal)
print(tcritunequal)
print(pvalunequal) 
```
