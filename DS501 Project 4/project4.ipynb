```python
# DS501 Project #4: Predictive Modeling Competition

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import mean_squared_error, f1_score, accuracy_score, classification_report
from sklearn.tree import DecisionTreeRegressor, DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestRegressor, RandomForestClassifier
from sklearn.neighbors import KNeighborsRegressor, KNeighborsClassifier
from sklearn.linear_model import LinearRegression, SGDRegressor, LogisticRegression
from sklearn.svm import SVC
from sklearn.impute import SimpleImputer # Added missing import
import matplotlib.pyplot as plt
import seaborn as sns

# Suppress warnings in the notebook output
import warnings
warnings.filterwarnings('ignore')

# --- TASK 1: DATA PREPROCESSING AND EDA 

data_regression = pd.read_csv('AmesHousing.csv')
data_classification = pd.read_csv('defaultofcreditcardclients.csv')

# --- Helper Function for Data Preparation ---
def prepare_data(data, target_column, is_regression=True):
    # Separate target variable
    X = data.drop(columns=[target_column])
    y = data[target_column]
    
    # Identify feature types (Placeholder - adjust columns based on actual data)
    numerical_features = X.select_dtypes(include=np.number).columns.tolist()
    categorical_features = X.select_dtypes(include='object').columns.tolist()
    
    # Create preprocessing pipelines
    numerical_transformer = Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='mean')),
        # MANDATORY for KNN and SVM (Task 1, Part 2)
        ('scaler', StandardScaler()) 
    ])
    
    categorical_transformer = Pipeline(steps=[
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('onehot', OneHotEncoder(handle_unknown='ignore'))
    ])
    
    # Create column transformer
    preprocessor = ColumnTransformer(
        transformers=[
            ('num', numerical_transformer, numerical_features),
            ('cat', categorical_transformer, categorical_features)
        ],
        remainder='passthrough' # Keep any unlisted columns
    )
        
    return X, y, preprocessor


# UNCOMMENT AND EXECUTE THESE BLOCKS ONCE YOUR DATA IS LOADED.

# --- TASK 1.1: REGRESSION SETUP (Ames Housing) ---
print("--- Starting Regression Task Setup (Ames Housing) ---")
X_reg, y_reg, preprocessor_reg = prepare_data(data_regression, 'SalePrice')
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
X_reg, y_reg, test_size=0.5, random_state=42) # 50/50 Split (Task 3, Part 1)

# --- TASK 1.2: CLASSIFICATION SETUP (Credit Card Default) ---
print("--- Starting Classification Task Setup (Credit Card Default) ---")
X_clf, y_clf, preprocessor_clf = prepare_data(data_classification, 'default_payment_next_month', is_regression=False)
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(X_clf, y_clf, test_size=0.5, random_state=42) # 50/50 Split (Task 3, Part 1)

```

```python
#Task 2 - Regression Task
#Linear - Analytical solution
#Building pipeline for training the model
Linear_reg_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_reg),
    ('model', LinearRegression())
])
#Fitting the data using the training input and output
Linear_reg_pipeline.fit(X_train_reg,y_train_reg)
#Predict the testing input to get the testing output
y_pred_lregression = Linear_reg_pipeline.predict(X_test_reg)
```

```python
#Linear - Gradient Descent
#Building pipeline for training the model
Gradian_Descent_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_reg),
    ('model', SGDRegressor())
])
#Fitting the data using the training input and output
Gradian_Descent_pipeline.fit(X_train_reg,y_train_reg)
#Predict the testing input to get the testing output
y_pred_GD = Gradian_Descent_pipeline.predict (X_test_reg)
```

```python
#Tree-Based - Decision Tree Regressor 
#Building pipeline for training the model
Decision_Tree_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_reg), 
    ('model', DecisionTreeRegressor())
])
#Fitting the data using the training input and output
Decision_Tree_pipeline.fit(X_train_reg,y_train_reg)
#Predict the testing input to get the testing output
y_pred_dt = Decision_Tree_pipeline.predict (X_test_reg)
```

```python
#Tree-Based - Random Forest 
#Building pipeline for training the model
Random_Forest_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_reg), 
    ('model', RandomForestRegressor())
])
#Fitting the data using the training input and output
Random_Forest_pipeline.fit(X_train_reg,y_train_reg)
#Predict the testing input to get the testing output
y_pred_rf = Random_Forest_pipeline.predict (X_test_reg)
```

```python
#Instance-Based-K-Nearest Neighbors 
#Building pipeline for training the model
KNearest_Neighbors_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_reg), 
    ('model', KNeighborsRegressor())
])
#Fitting the data using the training input and output
KNearest_Neighbors_pipeline.fit(X_train_reg,y_train_reg)
#Predict the testing input to get the testing output
y_pred_knn = KNearest_Neighbors_pipeline.predict (X_test_reg)
```

```python
#Linear - Linear Classification
#Building pipeline for training the model
Reg_Clf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_clf),   
    ('model', LogisticRegression())
])
#Fitting the data using the training input and output
Reg_Clf_pipeline.fit(X_train_clf, y_train_clf)
#Predict the testing input to get the testing output
y_pred_Reg_Clf = Reg_Clf_pipeline.predict(X_test_clf)
```

```python
#Tree Based - Decision Tree Classifier
#Building pipeline for training the model
DT_Clf_pipeline = Pipeline (steps=[
    ('preprocessor', preprocessor_clf),   
    ('model', DecisionTreeClassifier())
])
#Fitting the data using the training input and output
DT_Clf_pipeline.fit(X_train_clf, y_train_clf)
y_pred_DT_Clf = DT_Clf_pipeline.predict(X_test_clf)
```

```python
#Tree Based - Random Forest Classifier
#Building pipeline for training the model
RF_Clf_pipeline = Pipeline (steps=[
    ('preprocessor', preprocessor_clf),   
    ('model', RandomForestClassifier())
])
#Fitting the data using the training input and output
RF_Clf_pipeline.fit(X_train_clf, y_train_clf)
#Predict the testing input to get the testing output
y_pred_RF_Clf = RF_Clf_pipeline.predict(X_test_clf)
```

```python
#Instance Based - K-Nearest Neighbors Classifier
#Building pipeline for training the model
KNN_Clf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_clf), 
    ('model', KNeighborsClassifier())
])
#Fitting the data using the training input and output
KNN_Clf_pipeline.fit(X_train_clf, y_train_clf)
#Predict the testing input to get the testing output
y_pred_knn_Clf = KNN_Clf_pipeline.predict (X_test_clf)
```

```python
#Kernel Method - Supported Vector Machine Classfier
#Building pipeline for training the model
SVM_Clif_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor_clf),   
    ('model', SVC())
])
#Fitting the data using the training input and output
SVM_Clif_pipeline.fit(X_train_clf, y_train_clf)
#Predict the testing input to get the testing output
y_pred_svm_Clf = SVM_Clif_pipeline.predict (X_test_clf)
```
